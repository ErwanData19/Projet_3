Le client nous a informé qu'il y avait des valeurs à 0 pour les marqueurs "concavity" et souhaite que nous supprimions les lignes dans lesquelles ils apparaissent

Nous allons commencer par les identifier puis nous les supprimerons

```python
# identification des lignes avec des valeurs 0 pour les marqueurs "concavity"

condition = (df['concavity_mean'] == 0) | (df['concavity_se'] == 0) | (df['concavity_worst'] == 0)
rows_with_zeros = df[condition]

print(rows_with_zeros)
```

```python
# suppression des lignes

df_CduS = df_CduS[(df_CduS["concavity_mean"] != 0) & (df_CduS["concavity_se"] != 0) & (df_CduS["concavity_worst"] != 0)]
```

```python
df_CduS.shape
```

#Outliers

```python
print(df_CduS.head().to_markdown())
```

```python
# Séparer les données en fonction des étiquettes
df_benign = df_CduS[df_CduS['diagnosis'] == "B"]
df_malignant = df_CduS[df_CduS['diagnosis'] == "M"]

B_numeric = df_benign.select_dtypes(include = "number")
M_numeric = df_malignant.select_dtypes(include = "number")
```

##Outliers Cas Bénins

```python
# Détection des outliers pour les cas bénins
Q1_benign = B_numeric.quantile(0.25)
Q3_benign = M_numeric.quantile(0.75)
IQR_benign = Q3_benign - Q1_benign
lower_bound_benign = Q1_benign - 1.5 * IQR_benign
upper_bound_benign = Q3_benign + 1.5 * IQR_benign
outliers_benign = ((B_numeric < lower_bound_benign) | (B_numeric > upper_bound_benign)).any(axis=1)
outliers_df_benign = B_numeric[outliers_benign]
```

```python
print("Outliers dans les cas bénins:")
print(outliers_df_benign.to_markdown())
```

##Outliers Cas Malins

```python
# Détection des outliers pour les cas malins
Q1_malignant = M_numeric.quantile(0.25)
Q3_malignant = M_numeric.quantile(0.75)
IQR_malignant = Q3_malignant - Q1_malignant
lower_bound_malignant = Q1_malignant - 1.5 * IQR_malignant
upper_bound_malignant = Q3_malignant + 1.5 * IQR_malignant
outliers_malignant = ((M_numeric < lower_bound_malignant) | (M_numeric > upper_bound_malignant)).any(axis=1)
outliers_df_malignant = M_numeric[outliers_malignant]
```

```python
print("Outliers dans les cas malins:")
print(outliers_df_malignant.to_markdown())
```

##Total Outliers

```python
out_benign = pd.DataFrame(outliers_df_benign.nunique())
out_benign.rename(columns = {0: "Outliers B"}, inplace = True)

out_malign = pd.DataFrame(outliers_df_malignant.nunique())
out_malign.rename(columns = {0: "Outliers M"}, inplace = True)
```

```python
df_outliers = pd.concat([out_benign, out_malign], axis = 1)
df_outliers
```
